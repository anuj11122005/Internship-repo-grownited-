```python
import pandas as pd 
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

np.random.seed(42)
```

```python
months = ["Jan","Feb","Mar","Apr","May","Jun","Jul","Aug","Sep","Oct","Nov","Dec"]

categories = [
    "Electronics",
    "Furniture",
    "Grocery",
    "Clothing",
    "Sports"
]

rows = 300

df = pd.DataFrame({
    "Invoice": np.arange(1001,1001+rows),
    "Month": np.random.choice(months,rows),
    "Category": np.random.choice(categories,rows),
    "Revenue": np.random.randint(5000,80000,rows),
    "Quantity": np.random.randint(1,40,rows),
    "Discount": np.random.randint(0,31,rows),
    "Profit": np.random.randint(1000,20000,rows)
})

df.head()
```

```python
df.to_csv("retail_sales.csv",index=False)
print ("Dataset stored Successfully")
```

```python
sales = pd.read_csv("retail_sales.csv")
sales.head()
```

```python
sales.info()
```

```python
sales.isnull().sum()
```

```python
print("Total Records :", len(sales))
print("Total Columns :", len(sales.columns))

print("\nColumns")
print(sales.columns)

print("\nData Types")
print(sales.dtypes)
```

```python
sales.describe()
```

```python
total_revenue = sales["Revenue"].sum()
print("Total Revenue is :",total_revenue)
```

```python
average_revenue = sales["Revenue"].mean()
print("The Average Revenue :", round(average_revenue))
```

```python
median_revenue = sales["Revenue"].median()

print("Median Revenue :", median_revenue)
```

```python
mode_quantity = sales["Quantity"].mode()

print("Mode of Quantity :")
print(mode_quantity)
```

```python
minimum_revenue = sales["Revenue"].min()
maximum_revenue = sales["Revenue"].max()

print("Minimum Revenue :", minimum_revenue)
print("Maximum Revenue :", maximum_revenue)
```

```python
revenue_range = maximum_revenue - minimum_revenue

print("Revenue Range :", revenue_range)
```

```python
variance_profit = sales["Profit"].var()

print("Variance of Profit :", round(variance_profit, 2))
```

```python
std_revenue = sales["Revenue"].std()
print("Standard Deviation of Revenue :", round(std_revenue, 2))
```

```python
q1 = sales["Revenue"].quantile(0.25)
q2 = sales["Revenue"].quantile(0.50)
q3 = sales["Revenue"].quantile(0.75)

print("First Quartile (Q1) :", q1)
print("Second Quartile (Median) :", q2)
print("Third Quartile (Q3) :", q3)
```

```python
p10 = sales["Revenue"].quantile(0.10)
p25 = sales["Revenue"].quantile(0.25)
p50 = sales["Revenue"].quantile(0.50)
p75 = sales["Revenue"].quantile(0.75)
p90 = sales["Revenue"].quantile(0.90)

print("10th Percentile :", p10)
print("25th Percentile :", p25)
print("50th Percentile :", p50)
print("75th Percentile :", p75)
print("90th Percentile :", p90)
```

```python
iqr = q3 - q1

print("Interquartile Range (IQR) :", q3 - q1)
```
