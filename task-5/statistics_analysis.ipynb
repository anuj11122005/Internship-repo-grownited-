```python
import pandas as pd 
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

np.random.seed(42)
```

```python
months = ["Jan","Feb","Mar","Apr","May","Jun","Jul","Aug","Sep","Oct","Nov","Dec"]

categories = [
    "Electronics",
    "Furniture",
    "Grocery",
    "Clothing",
    "Sports"
]

rows = 300

df = pd.DataFrame({
    "Invoice": np.arange(1001,1001+rows),
    "Month": np.random.choice(months,rows),
    "Category": np.random.choice(categories,rows),
    "Revenue": np.random.randint(5000,80000,rows),
    "Quantity": np.random.randint(1,40,rows),
    "Discount": np.random.randint(0,31,rows),
    "Profit": np.random.randint(1000,20000,rows)
})

df.head()
```

```python
df.to_csv("retail_sales.csv",index=False)
print ("Dataset stored Successfully")
```
