```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

plt.style.use("ggplot")
```

```python
players = pd.read_csv("ipl_players.csv")
matches = pd.read_csv("ipl_matches.csv")
```

```python
print("IPL Players Dataset")
display(players)

print("\nIPL Matches Dataset")
display(matches)
```

```python
print("Players Dataset Information")
players.info()

print("\nMatches Dataset Information")
matches.info()
```

```python
print("Players Missing Values")
print(players.isnull().sum())

print("\nMatches Missing Values")
print(matches.isnull().sum())
```

# Task 1: Display Top 5 Run Scorers

```python
top_runs = players.sort_values(by="Runs", ascending=False).head(5)

print("Top 5 Run Scorers")
print(top_runs[["Player", "Runs"]])
```

# Task 2: Display Top 5 Wicket Takers

```python
top_wickets = players.sort_values(by="Wickets", ascending=False).head(5)

print("Top 5 Wicket Takers")
print(top_wickets[["Player", "Wickets"]])
```

# Task 3: Display Team Points Table

```python
points_table = matches.sort_values(by="Points", ascending=False)

print("Team Points Table")
print(points_table[["Team", "Points"]])
```

# Task 4: Find Player with Highest Strike Rate

```python
highest_sr = players.loc[players["StrikeRate"].idxmax()]

print("Highest Strike Rate")
print("Player :", highest_sr["Player"])
print("Strike Rate :", highest_sr["StrikeRate"])
```

# Task 5: Find Team with Highest Points

```python
best_team = matches.loc[matches["Points"].idxmax()]

print("Best Team")
print("Team :", best_team["Team"])
print("Points :", best_team["Points"])
```

# Task 6: Calculate Average Runs

```python
average_runs = players["Runs"].mean()

print("Average Runs :", round(average_runs, 1))
```

# Task 7: Players Having More Than 500 Runs

```python
players_500 = players[players["Runs"] > 500]

print("Players Having More Than 500 Runs")
print(players_500[["Player", "Runs"]])
```

# Task 8: Find All-Rounders

```python
all_rounders = players[
    (players["Runs"] > 300) &
    (players["Wickets"] > 10)
]

print("All Rounders")
print(all_rounders[["Player", "Runs", "Wickets"]])
```

# Chart 1: Top 5 Run Scorers (Bar Chart)

```python
top_runs = players.sort_values(by="Runs", ascending=False).head(5)

plt.figure(figsize=(8,5))
plt.bar(top_runs["Player"], top_runs["Runs"])

plt.title("Top 5 Run Scorers")
plt.xlabel("Players")
plt.ylabel("Runs")
plt.xticks(rotation=20)

plt.show()
```

# Chart 2: Top 5 Wicket Takers (Bar Chart)

```python
top_wickets = players.sort_values(by="Wickets", ascending=False).head(5)

plt.figure(figsize=(8,5))
plt.bar(top_wickets["Player"], top_wickets["Wickets"])

plt.title("Top 5 Wicket Takers")
plt.xlabel("Players")
plt.ylabel("Wickets")
plt.xticks(rotation=20)

plt.show()
```

# Chart 3: Team Points Table (Bar Chart)

```python
points = matches.sort_values(by="Points", ascending=False)

plt.figure(figsize=(8,5))
plt.bar(points["Team"], points["Points"])

plt.title("IPL Points Table")
plt.xlabel("Teams")
plt.ylabel("Points")

plt.show()
```

# Chart 4: Runs Distribution (Histogram)

```python
plt.figure(figsize=(8,5))

plt.hist(players["Runs"], bins=5)

plt.title("Runs Distribution")
plt.xlabel("Runs")
plt.ylabel("Frequency")

plt.show()
```

# Chart 5: Strike Rate vs Runs (Scatter Plot)

```python
plt.figure(figsize=(8,5))

plt.scatter(players["StrikeRate"], players["Runs"])

plt.title("Strike Rate vs Runs")
plt.xlabel("Strike Rate")
plt.ylabel("Runs")

plt.show()
```

# Chart 6: Team Winning Percentage (Pie Chart)

```python
matches["WinningPercentage"] = (
    matches["Won"] / matches["Matches"]
) * 100
```

```python
plt.figure(figsize=(7,7))

plt.pie(
    matches["WinningPercentage"],
    labels=matches["Team"],
    autopct="%1.1f%%"
)

plt.title("Team Winning Percentage")

plt.show()
```

# Chart 7: Correlation Heatmap

```python
plt.figure(figsize=(8,6))

sns.heatmap(
    players[
        ["Runs","Wickets","Sixes","Fours","StrikeRate"]
    ].corr(),
    annot=True,
    cmap="coolwarm"
)

plt.title("Correlation Heatmap")

plt.show()
```

# Dashboard Layout

```python
import matplotlib.pyplot as plt
import seaborn as sns

top_runs = players.sort_values(by="Runs", ascending=False).head(5)
top_wickets = players.sort_values(by="Wickets", ascending=False).head(5)
points_table = matches.sort_values(by="Points", ascending=False)

matches["WinningPercentage"] = (matches["Won"] / matches["Matches"]) * 100

fig, axes = plt.subplots(3, 2, figsize=(18, 15))

# Chart 1: Top Run Scorers
axes[0,0].bar(top_runs["Player"], top_runs["Runs"])
axes[0,0].set_title("Top 5 Run Scorers")
axes[0,0].tick_params(axis='x', rotation=20)

# Chart 2: Top Wicket Takers
axes[0,1].bar(top_wickets["Player"], top_wickets["Wickets"])
axes[0,1].set_title("Top 5 Wicket Takers")
axes[0,1].tick_params(axis='x', rotation=20)

# Chart 3: Team Points
axes[1,0].bar(points_table["Team"], points_table["Points"])
axes[1,0].set_title("IPL Points Table")

# Chart 4: Winning Percentage
axes[1,1].pie(
    matches["WinningPercentage"],
    labels=matches["Team"],
    autopct="%1.1f%%",
    startangle=90
)
axes[1,1].set_title("Winning Percentage")

# Chart 5: Strike Rate vs Runs
axes[2,0].scatter(players["StrikeRate"], players["Runs"])
axes[2,0].set_title("Strike Rate vs Runs")
axes[2,0].set_xlabel("Strike Rate")
axes[2,0].set_ylabel("Runs")

# Chart 6: Heatmap
sns.heatmap(
    players[["Runs","Wickets","Sixes","Fours","StrikeRate"]].corr(),
    annot=True,
    cmap="coolwarm",
    ax=axes[2,1]
)
axes[2,1].set_title("Correlation Heatmap")

plt.suptitle("IPL DATA ANALYTICS DASHBOARD", fontsize=20)

plt.tight_layout()

plt.savefig("dashboard.png")

plt.show()
```

# Final IPL Analytics Report

```python
print("=" * 45)
print("        IPL ANALYTICS REPORT")
print("=" * 45)

print("Total Players :", len(players))

top_scorer = players.loc[players["Runs"].idxmax()]
print("\nTop Scorer :", top_scorer["Player"])
print("Runs :", top_scorer["Runs"])

top_bowler = players.loc[players["Wickets"].idxmax()]
print("\nTop Bowler :", top_bowler["Player"])
print("Wickets :", top_bowler["Wickets"])

best_team = matches.loc[matches["Points"].idxmax()]
print("\nBest Team :", best_team["Team"])
print("Points :", best_team["Points"])

highest_sr = players.loc[players["StrikeRate"].idxmax()]
print("\nHighest Strike Rate :")
print(highest_sr["Player"])
print("Strike Rate :", highest_sr["StrikeRate"])

print("\nAll-Rounders :")
for player in all_rounders["Player"]:
    print(player)

print("\nAverage Runs :")
print(round(players["Runs"].mean(), 1))

print("=" * 45)
print("END OF REPORT")
print("=" * 45)
```
