```python
import matplotlib.pyplot as plt 
```

```python
months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]
sales = [12000, 15000, 18000, 14000, 22000, 25000]
```

```python
plt.plot(months, sales, marker="o")

plt.title("Monthly Sales Dashboard - Line Chart")
plt.xlabel("Months")
plt.ylabel("Sales")

plt.grid()
plt.show()
```

```python
plt.bar(months, sales, color="red")

plt.title("Monthly Sales Dashbord - Bar Chart")
plt.xlabel("Months")
plt.ylabel("Sales")

plt.show()
```

```python
plt.fill_between(months,sales)

plt.title("Monthly Sales Dashboard - Area Chart")
plt.xlabel("Month")
plt.ylabel("Sales")

plt.grid()

plt.show()
```
