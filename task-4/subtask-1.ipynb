```python
import matplotlib.pyplot as plt
```

```python
names = ["Amit", "Neha", "Rahul", "Priya", "Jay"]
marks = [70, 92, 88, 65, 45]
```

```python
plt.bar(names, marks)
plt.title("Student marks - Bar chart")
plt.xlabel("Student")
plt.ylabel("Marks")
plt.show()
```

```python
plt.pie(marks, labels=names, autopct="%1.1f%%")
plt.title("Student Marks - Pie Chart")
plt.show()
```

```python
plt.plot(names, marks, marker="o")
plt.title("Student Marks - Line Chart")
plt.xlabel("Students")
plt.ylabel("Marks")
plt.grid()
plt.show()
```
