```python
import matplotlib.pyplot as plt
```

```python
employees = ["A", "B", "C", "D", "E"]
salary = [30000, 45000, 55000, 70000, 50000]
```

```python
plt.hist(salary, bins=20)

plt.title("Employees Salary Dashboard - Histogram Chart")
plt.xlabel("Salary")
plt.ylabel("Frequency")

plt.show()
```

```python
plt.boxplot(salary)

plt.title("Employees Salary Box plot")
plt.xlabel("Salary")

plt.show()
```

```python
plt.bar(employees, salary, color="yellow")

plt.title("Employees Salary Dashboard - Bar Chart")
plt.xlabel("Employees")
plt.ylabel("Salary")

plt.show()
```
